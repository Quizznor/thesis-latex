```python
n_points = 20
r = 1

angles = np.linspace(0, 2*np.pi, n_points)
x = r * np.cos(angles)
y = r * np.sin(angles)

x += np.random.normal(0, r * 0.1, n_points)
y += np.random.normal(0, r * 0.1, n_points)

plt.plot(x, y)
plt.gca().set_aspect('equal')

for xx, yy in zip(x, y):
    print(f"({xx:.3f},{yy:.3f})")
```

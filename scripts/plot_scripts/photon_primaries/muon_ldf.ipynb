```python
from utils.binaries import *
from utils.plotting import *
```

```python
path = "/cr/tempdata01/filip/json_data/icrc2025-test7/"
df_proton = tools.pickle_load(f"{path}/proton.pkl")
df_photon = tools.pickle_load(f"{path}/photon.pkl")
```

```python
set_plt_style("double")

lgE = np.log10(df_proton["shower_size"])
lin_scale = (lgE - lgE.min()) / (lgE.max() - lgE.min()) 
colors = plt.cm.plasma(lin_scale.values)

plt.scatter(df_proton["spd"], df_proton["muon_density"], s=0.1, ec='none', fc=colors)
# plt.yscale("log")
plt.xlabel("$r$ / m")
plt.ylabel(r"$\rho^\mathrm{rec.}_\mu$ / $\mathrm{m}^{-2}$")

cbar = plot.add_colorbar(np.arange(16, 18.51, 0.05), plt.cm.plasma, label="$\log_{10}(E\,/\,\mathrm{eV})$")
cbar.set_ticks(np.linspace(16, 18.5, 6))
cbar.set_ticklabels(np.linspace(16, 18.5, 6))

plot.save(plt.gcf(), "photon_proton")
```

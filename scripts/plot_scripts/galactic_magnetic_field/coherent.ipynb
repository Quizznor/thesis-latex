```python
x, y, z, Bx, By, Bz = np.loadtxt("MichaelUnger-GalacticMagneticFieldUF23-a66bb6c/field.txt", unpack=True)

x_unique, y_unique = np.unique(x), np.unique(y)

X, Y = np.meshgrid(x_unique, y_unique)
Bx = Bx.reshape(len(X), len(Y))
By = By.reshape(len(X), len(Y))
```

```python
B_xy = np.sqrt(Bx**2 + By**2)
v_max = np.max(B_xy)

field = plt.imshow(B_xy, 
                   extent=[
                       np.min(x), np.max(x), np.min(y), np.max(x)
                    ],
                    cmap=plt.cm.plasma, 
                    vmin=0, vmax=v_max)

# field = plt.pcolormesh(x, y, B_xy)
plt.scatter(-8, #8 * np.sin(30 * np.pi/180),
            0, #8 * np.cos(30 * np.pi/180),
            marker='*', ec='k', fc='w', lw=0.15, s=20)

plt.xlabel("x / kpc")
plt.ylabel("y / kpc")

plt.gcf().colorbar(field, label=r"$|\vec{B}_{xy}|$ / $\upmu$G")
```

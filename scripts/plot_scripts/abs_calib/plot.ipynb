```python
olo_x, olo_y = np.loadtxt("olomouc.csv", delimiter=',', unpack=True)
```

```python
from scipy.optimize import curve_fit

def inverse_square(x, A):
    # return A/x**2
    return A*np.exp(-x)

popt_olo, pcov_olo = curve_fit(inverse_square, olo_x, olo_y * 1e9)
```

```python
plt.scatter([], [])
plt.plot([], [])

X = np.linspace(10, 560, 500)

plt.scatter(olo_x, olo_y * 1e9, label="Olomouc")
plt.plot(X, inverse_square(X, *popt_olo), marker='none')
# plt.plot(X, inverse_square(X, 5), marker='none')


plt.ylabel(r"Diode current / nA")
plt.xlabel(r"Diode distance / mm")
plt.legend()
# plt.yscale("log")
```
